This week we turn to vision and its role in how we perceive and make sense of sound and music. Visual cues colour how we interpret a sound, set up expectations, and help us follow a complex performance. A musician's sound-producing actions or a conductor's gestures influence how you hear rhythm and melody and may change your emotional response. We will look at how the brain combines what we see with what we hear, work through auditory–visual illusions, and consider why visual feedback matters when learning and performing music.


## Auditory–visual interaction

This chapter first clarifies how vision and audition interact, especially in music psychology and technology, before we turn to the anatomy of the eye and the use of [eye tracking](https://en.wikipedia.org/wiki/Eye_tracking) in performance and perception research.


### Auditory–visual and audio–video

In this course we distinguish between the terms *auditory–visual* and *audio–video*. They refer to different domains:

- **Audio–video**: These terms relate to the technical capture, processing, and reproduction of sound and moving images. [Audio](https://en.wikipedia.org/wiki/Audio) refers to the recording, transmission, and playback of sound signals, whether analogue or digital. [Video](https://en.wikipedia.org/wiki/Video) involves the recording, processing, and display of moving images.

- **Auditory–visual**: These terms denote the senses and their perceptual processes. The [auditory system](https://en.wikipedia.org/wiki/Auditory_system) transduces air-pressure fluctuations (sound) into neural signals that encode pitch, loudness, timbre and spatial position. The [visual system](https://en.wikipedia.org/wiki/Visual_system) transduces visible electromagnetic radiation (light) into neural representations of luminance, colour, form, motion and depth. Perception arises from both the physical properties of stimuli (frequency/wavelength, intensity) and neural processing (adaptation, attention, context), and it underlies how we integrate cues across the senses.

Keeping these apart helps distinguish the physical phenomena (sound and light), how we perceive them (audition and vision), and how we capture them (audio and video). The distinction matters for understanding how audio–video systems work and how we perceive our surroundings, and it matters especially in music, where seeing and hearing combine to shape the experience.

```{admonition} Why "audiovisual" is a tricky word
:class: note
The everyday word *audiovisual* takes one half from each pair: *audio* is the capture word, *visual* the perception word. It therefore straddles the two domains above. An "audiovisual recording" is a statement about files; "audiovisual integration" is a statement about what a brain does with what the eyes and ears report. The same word can cover a camera setting and a perceptual effect. That is why we try to avoid the term in this book and instead use *audio–video* for the recording and *auditory–visual* for the perception.
```

### Integration of the senses

Auditory–visual integration is the brain's ability to combine sound and sight to sharpen perception and comprehension; it underpins tasks like speech perception and shapes multimedia experiences.

Two terms are easy to confuse here. *[Multimodal perception](https://en.wikipedia.org/wiki/Multimodal_perception)* is the ordinary case: several senses contribute to one percept at the same time. When you watch and hear a drummer, you perceive one stroke, not a sight plus a sound, and each sense adds information the other lacks. The eyes give the preparation and the place of the stroke, the ears its timing and timbre. The senses are combined, and the result is richer than either alone. *[Crossmodal perception](https://en.wikipedia.org/wiki/Crossmodal)*, on the other hand, is when one sense changes what another sense reports: what you see alters what you hear, or what you hear alters what you see, although the information reaching the second sense has not changed. 

Multimodal and crossmodal perception are not rival theories but two views of the same machinery. They are both based on sensory integration. A crossmodal effect is what that integration looks like when the senses disagree and the brain settles the disagreement in favour of one of them. The result is often an illusion.

Crossmodal effects range from the bouba–kiki pairing of sounds with shapes met in [tuning in](tuning-in.ipynb) to illusions in which one sense overrides another. One famous auditory–visual illusion is the [McGurk effect](https://en.wikipedia.org/wiki/McGurk_effect).

<iframe width="560" height="315" src="https://www.youtube.com/embed/2k8fHR9jKVM?si=KVW2W2XM8fIrcPtR" title="YouTube video player" frameborder="0" allow="accelerometer; autoplay; clipboard-write; encrypted-media; gyroscope; picture-in-picture; web-share" referrerpolicy="strict-origin-when-cross-origin" allowfullscreen></iframe>

When the visible mouth movements of one speech sound are paired with the acoustic signal of another, observers often perceive a third, fused sound (classically: audio "ba" + visual "ga" → perceived "da"). Vision can change what we hear at a basic perceptual level, not only our later judgements.

:::{note} Dig deeper: what drives and modulates the McGurk effect
:class: dropdown
The effect arises because the brain combines temporally coincident but conflicting cues from the two senses within a limited time window; cortical areas such as the superior temporal sulcus, a region where auditory and visual information meet, are implicated in resolving the mismatch. Several factors modulate it: timing (asynchrony reduces fusion), signal clarity (degraded audio strengthens visual influence), attention, and individual differences (e.g., lip-reading skill).
:::

The same principles explain why, in music, seeing a performer's gestures, facial movements, or instrument actions can change the perceived onset, articulation, or expressiveness of a sound. The effect has been shown for singing: when the video of a singer performing one interval was paired with the sound of a different one, viewers heard the interval as smaller or larger depending on what the face was doing {cite:p}`Laeng2021`.

Vision is not the only partner sense: touch and vibration also carry musical structure, from bass felt in the chest at a concert to vibrotactile interfaces, and those channels are treated in [the body](the-body.ipynb).


### Ventriloquism, spatial capture, and temporal binding

The *[ventriloquism effect](https://en.wikipedia.org/wiki/Ventriloquism_effect)* is named after the stage trick, but in perception research it means something broader. When a sound and a sight arrive close together in time, we hear the sound as coming from where we see its source. In a concert, a film, a game or a VR headset, where a sound seems to come from therefore depends both on the loudspeakers or headphones and on what you see. This is one reason why staging, lighting and picture cuts interact with the mix. The effect is strongest inside a short *binding window*, from tens to a few hundred milliseconds depending on the task. When sight and sound are clearly out of step, it fades.

The sketch below shows how such binding is usually pictured: two smooth, bell-shaped pulses of onset energy, one auditory and one visual, and a shaded band marking the rough time window within which the brain tends to bind the two into a single event. The widths are illustrative rather than fitted to individual perceivers.


````python
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams.update({
    "font.size": 10,
    "axes.titlesize": 11,
    "axes.labelsize": 10,
    "xtick.labelsize": 9,
    "ytick.labelsize": 9,
    "legend.fontsize": 9,
    "figure.figsize": (10, 4),
    "figure.dpi": 100,
    "savefig.dpi": 150,
    "axes.spines.top": False,
    "axes.spines.right": False,
})

t = np.linspace(-0.25, 0.55, 800)
audio = np.exp(-((t - 0.0) ** 2) / (2 * 0.012**2))
visual = np.exp(-((t - 0.07) ** 2) / (2 * 0.018**2))

fig, ax = plt.subplots(figsize=(10, 3))
ax.plot(t * 1000, audio, label="Auditory onset energy (model)")
ax.plot(t * 1000, visual, label="Visual onset energy (model)")
ax.axvspan(-120, 120, alpha=0.15, color="green", label="±120 ms binding window (sketch)")
ax.set_xlabel("Time relative to auditory onset (ms)")
ax.set_ylabel("Energy (arbitrary)")
ax.set_title("Schematic auditory–visual timing (illustrative)")
ax.legend(loc="upper right")
plt.tight_layout()
plt.show()
````

*Figure: A schematic of auditory–visual binding: model onset-energy pulses for a sound and a sight, and the shaded window of about ±120 ms within which the two tend to be bound into one event. The widths are illustrative.*

### Notation, score reading, and gaze

Many performance traditions ask you to split your attention between a score, a conductor, your co-performers and your own instrument. Eye-tracking studies of sight-reading show where the eyes rest. They linger on the places that matter for the structure, such as downbeats, cue notes and page turns, and often return to a spot near a phrase boundary {cite:p}`MadellHebert2008`. This connects to the ideas of prediction and subdivision in [time and rhythm](time-and-rhythm.ipynb).

:::{note} Dig deeper: how firm is the evidence on music reading?
:class: dropdown
A methodological review of over two decades of music-reading studies shows that tasks, measures, and stimuli vary so much that general conclusions remain tentative {cite:p}`Puurtinen2018`. Studies comparing paper and screen scores suggest differences in scrolling, peripheral preview, and markup, but here too the evidence base is still thin.
:::

### Judging with the eyes

If vision can change *what* we hear, it can also change what we think of a performance. The best-known demonstration comes from a series of experiments by Chia-Jung Tsay {cite:p}`Tsay2013`. Participants—musical novices and professional musicians alike—were asked to pick the winner of an international classical music competition from short excerpts of the finalists. Some received sound only, some video only, some both.

The result was the opposite of what almost everyone predicted. Participants given *silent video* identified the actual winners at rates well above chance. Those given the sound, with or without video, did not. Asked beforehand, more than four out of five participants said that sound was what mattered most to them; their own judgements said otherwise.

A large replication found the same pattern with the original clips but showed that it holds only under limited conditions. With other clips from comparable performances, judgements were at or below chance, and when the performers differed clearly in quality, participants were most accurate when they could hear the music {cite:p}`Mehr2018`. What survives is the core point: the visual channel is not decoration on top of an auditory art. It carries information about effort, confidence, expressive intent, and stagecraft, and perceivers use it whether or not they believe they do.

For musicology this cuts in several directions. It bears on how competitions and auditions are run. Orchestras now often run auditions where performers stand behind a screen to remove exactly this kind of bias. It also bears on how we should treat historical accounts of performances that no longer exist except as recordings. And it bears on our own analytic habits. An analysis based on a recording is an analysis of a deliberately reduced version of the event.

### The concert as a visual event

There are several practical implications of auditory–visual theory. For example, deciding on where performers stand on a stage relative to each other and to the audience sets both what you can see and what you can hear. A soloist stepping forward is a change in visual salience and in direct-to-reverberant ratio at once. Lighting can also be used to directly focus attention. A spotlight is an instruction about where to look, and therefore about which stream to follow.

The increasing use of screens in live concerts attests to the same. Large venues relay close-ups of performers' hands and faces. At many large concerts, most of the audience watches and listens to mediated images and amplified sound, sometimes arriving from different directions. Livestreamed and recorded concerts go further, since a director's cuts decide what the viewer attends to, moment by moment, an editorial listening guide that an acoustic concert does not impose.

The same logic shapes music in film, television, and games, where sound and image are constructed together. [Michel Chion](https://en.wikipedia.org/wiki/Michel_Chion)'s *synchresis*, met in [listening](listening.ipynb), is the working principle {cite:p}`Chion2019`. Pair a sound with a visible event and the two weld together, whatever the sound's real source. Foley artists, game audio designers, and composers all rely on it.

## The eye

This section gives a practical overview of the eye and why it matters for studies of sound, music, and performance.


### Anatomy and optics

The [human eye](https://en.wikipedia.org/wiki/Human_eye) is an optical organ that focuses light onto the [retina](https://en.wikipedia.org/wiki/Retina), its photosensitive inner surface. The eye adapts to changing light: in bright surroundings the pupil constricts to sharpen the image and reduce optical distortions; in dim light it dilates to maximise the number of photons reaching the retinal photoreceptors.

The [cornea](https://en.wikipedia.org/wiki/Cornea), the transparent front surface of the eye, does most of the focusing, while the [lens](https://en.wikipedia.org/wiki/Lens_(vertebrate_anatomy)), just behind the pupil, fine-tunes the focus so that images fall sharply on the retina.

The [iris](https://en.wikipedia.org/wiki/Iris_(anatomy)), a ring of pigmented muscle tissue, controls the size of the [pupil](https://en.wikipedia.org/wiki/Pupil). The pupil's diameter ranges from about 2 to 8 mm, yet natural light levels can vary by a factor of a million (for example, from moonlight to sunlight).

```{image} figures/vision/01-7fc903cd-b4f5-4d77-a1e7-98ccae56b9bf.jpg
:alt: Anatomy of the Eye.
:width: 100%
```

*Figure: The main parts of the human eye.*


The retina turns light into neural signals and does some processing of its own, such as sharpening contrast, finding edges and detecting motion. It has three kinds of light sensor. Rods are very sensitive, work in dim light, and see only in shades of grey. Cones work in bright light, give us colour, and are packed into the fovea. A third kind, the light-sensitive ganglion cells (ipRGCs), help set pupil size and the body's daily rhythm.

### Eye movements and the pupil

The eyes move in a few basic ways. [Saccades](https://en.wikipedia.org/wiki/Saccade) are very fast jumps that land the sharp centre of vision (the fovea) on a new spot, with [fixations](https://en.wikipedia.org/wiki/Fixation_%28visual%29) lasting around 200–400 ms. Smooth pursuit lets the eyes follow a moving object when you can see it. Vergence changes the angle between the two eyes so you can focus on near versus far things. In music settings, where someone looks—at a face, the hands, or the score—strongly shapes the visual information they pick up, and that input can change how they hear, understand, or perform the music.


Pupil size is set by the muscles of the iris, under control of the autonomic nervous system. The pupil trades light against sharpness: a small pupil gives a sharper image with more depth of field, a large pupil lets in more light. Pupil size also changes with states that have nothing to do with light, such as mental effort, surprise and emotional arousal. These changes are driven by the brain's arousal systems, and they can be measured with pupillometry. When reading pupil data, remember that lighting, task difficulty and emotional context all change pupil size.

## Eye tracking and pupillometry

The eye movements and pupil responses described above can be measured. Two methods do most of the work in music research: eye tracking, which records where the eyes point, and pupillometry, which records how wide the pupils are. Both are cameras pointed at the eyes, and both turn vision into data in the same way that motion capture turns movement into data in [the body](the-body.ipynb).

### Eye tracking

Eye tracking measures where and how the eyes move, which tells us about visual attention, perception, and the cognitive processes behind them. It is used in psychology, neuroscience, marketing, usability studies, and human–computer interaction.

There are two main types of eye tracker: mobile and stationary. Mobile eye trackers are wearable devices, such as glasses, that record eye movements in natural, real-world settings. They suit studies that involve movement, such as sports, navigation, or field research.

Stationary eye trackers are fixed devices, often mounted on a desk or built into a monitor, used in controlled laboratory settings. They suit experiments on reading, visual search, or website usability.

The three kinds of tracker met in this course make different compromises. The figures below are typical values; check the specification of the device you use.

| Tracker | Sampling rate | Accuracy | Freedom of movement | Cost |
|---|---|---|---|---|
| Mobile glasses | 100–200 Hz | about 1° | Walk, play, turn the head | High |
| Stationary (desk or chin rest) | 500–2000 Hz | below 0.5° | Head still in front of a screen | High |
| Laptop webcam | 30 Hz, the camera frame rate | several degrees | Head still, and it drifts | Free |

There is no right or wrong here. Mobile glasses buy freedom with precision, stationary trackers buy precision with freedom, and a webcam gives up both for access.

### Gaze tracking

[Gaze](https://en.wikipedia.org/wiki/Gaze_(physiology)) is the direction of a person's visual attention, and it is our main observable proxy for where cognitive resources go. In music research, gaze reveals what performers and perceivers attend to (hands, face, score, conductor), how they sample information over time, and how visual cues shape what they hear.

```{image} figures/vision/hero_f60f7de1-4cda-47f7-98bc-f5415956abde.JPG
:alt: Gaze tracking
:width: 100%
```

*Figure: Eye fixations of several individuals (coloured circles) and their durations (circle size). Image credit: Bruno Laeng/UiO.*

A tracker records the fixations and saccades described above. Fixation count and duration are common measures of interest or difficulty; saccade amplitude and direction reveal scanning strategies; and the sequence of fixations and saccades, the *scanpath*, describes a viewing strategy.

Gaze is a useful proxy for attention, but not a perfect one, since people can attend covertly without moving their eyes. Calibration, sampling rate, and tracker type (mobile vs stationary) set the accuracy and decide which metrics are reliable; for precise saccade dynamics you want a high sampling rate (e.g., >250 Hz). Define areas of interest (AOIs) carefully, using dynamic AOIs for moving performers, and account for head motion in mobile settings. Always read gaze patterns against the task demands, the participant's expertise, and the stimulus timing, for example the anticipatory fixations seen in sight-reading.


Eye-tracking output is a sequence of fixations (where the eyes paused, and for how long) joined by saccades. Below we simulate fixations sweeping across a notional score and show the two standard views: a *scanpath* (order and dwell time) and a *heatmap* (where attention accumulated).

````python
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams.update({
    "font.size": 10,
    "axes.titlesize": 11,
    "axes.labelsize": 10,
    "xtick.labelsize": 9,
    "ytick.labelsize": 9,
    "legend.fontsize": 9,
    "figure.figsize": (10, 4),
    "figure.dpi": 100,
    "savefig.dpi": 150,
    "axes.spines.top": False,
    "axes.spines.right": False,
})

rng = np.random.default_rng(2)
n = 25
xs = np.sort(rng.uniform(0, 10, n))                       # left-to-right reading
ys = 1.5 + 0.6 * np.sin(xs * 1.5) + 0.25 * rng.standard_normal(n)
dur = rng.uniform(0.15, 0.5, n)                           # fixation durations (s)

fig, ax = plt.subplots(2, 1, figsize=(10, 5))
ax[0].plot(xs, ys, "-", color="0.6", lw=0.8)
ax[0].scatter(xs, ys, s=dur * 800, alpha=0.5)
for i, (x, y) in enumerate(zip(xs, ys)):
    ax[0].text(x, y, str(i + 1), ha="center", va="center")
ax[0].set_title("Gaze scanpath over a notional score (circle size = fixation duration)")
ax[0].set_xlim(0, 10); ax[0].set_ylim(0, 3)

H, _, _ = np.histogram2d(xs, ys, bins=[20, 8], range=[[0, 10], [0, 3]], weights=dur)
ax[1].imshow(H.T, origin="lower", aspect="auto", extent=[0, 10, 0, 3], cmap="hot")
ax[1].set_title("Fixation-duration heatmap")
ax[1].set_xlabel("Horizontal position")
plt.tight_layout()
plt.show()
````

*Figure: Simulated eye-tracking data over a notional score. Top: the scanpath, fixations numbered in order with circle size showing dwell time. Bottom: the same fixations as a heatmap of accumulated dwell time.*

The [Eye tracker](https://fourms.github.io/sensingsoundandmusic/apps/eye-tracker/) app shows the principle with a laptop webcam. It finds the dark pupil in the camera image, calibrates with nine points on the screen, and then draws where you look on a page of notation, a text, or a grid of shapes, as a live scanpath and a heatmap, while plotting the area of each pupil over time. ### Pupillometry

[Pupillometry](https://en.wikipedia.org/wiki/Pupillometry) is the measurement of pupil size and how it changes over time. Because the pupil responds not only to light but also to cognitive and emotional states, pupillometry lets us study mental effort, arousal, and attention.

```{image} figures/vision/hero_7c46327d-c031-442b-81dc-32761f3c91eb.jpg
:alt: Pupillometry example
:width: 100%
```

*Figure: Checking the pupil size before [MusicLab Copenhagen](https://www.uio.no/ritmo/english/projects/musiclab/2021/dsq/index.html).*

Decades of research show that pupil diameter tracks mental workload. In a classic study, {cite:t}`HessPolt1964` found that the pupil widened step by step as participants solved harder and harder arithmetic problems. [Daniel Kahneman](https://en.wikipedia.org/wiki/Daniel_Kahneman), later known for *[Thinking, Fast and Slow](https://en.wikipedia.org/wiki/Thinking,_Fast_and_Slow)* (2011), showed that the pupil also widens with each extra digit held in memory, and narrows again as the digits are reported {cite:p}`KahnemanBeatty1966`. Hundreds of later studies have confirmed this: a larger pupil, or a brief dilation, goes with greater cognitive and attentional demands. Music gives the clearest emotional example. In the chills study described in [physiology](physiology.ipynb), the pupil widened at the moments where perceivers reported a shiver, most of all for music they had chosen themselves {cite:p}`Laeng2016`. A camera pointed at the eye can thus time an emotion that the person only reports afterwards.

### Eye tracking and pupillometry in music research

As the equipment has become cheaper and easier to use, eye tracking has spread through music research. Some typical uses:

- **Watching a concert:** which gestures and movements the audience looks at, and how what they see shapes how expressive the music seems.
- **Reading music:** how musicians' fixations and saccades across a score differ with expertise and sight-reading skill.
- **Auditory–visual illusions:** experiments such as the McGurk effect, where sound and picture disagree and perception follows a mixture of both.
- **Emotion:** pupil size as a window on arousal while perceivers hear emotionally charged music.
- **Performance anxiety:** how gaze and pupil change under stress on stage, as markers of cognitive load and attention.
- **Interactive installations:** where a participant looks controls or triggers the sound.

In all of these, eye tracking and pupillometry show how seeing and hearing work together.

### Gaze in performance: conductors and ensembles

Performers use their eyes to cue an entry, mark a phrase boundary and stay together. Ensemble players tend to glance up at structural boundaries and risky transitions. A conductor's gaze spreads attention across the players and signals who comes next. Gaze is part of the visual channel that runs alongside the sound when people make music together, which [the body](the-body.ipynb) chapter approaches from the side of movement.

## A critical look: do blind people hear better?

It is a common belief, reinforced by fiction and by the fame of blind musicians, that losing sight sharpens the ears. This chapter has argued that hearing and seeing are intertwined, so the claim is a natural test case.

- **The claim** is that blindness makes the ears more sensitive, and that blind musicians excel because they hear more than the rest of us.
- **The evidence** says something more specific. Hearing thresholds are no better in blind people; the ears are the same. What differs is what the brain does with the signal. People blind from early in life locate sounds in the horizontal plane more accurately than sighted people, particularly when listening with one ear {cite:p}`Lessard1998`, and they detect changes in pitch faster and more finely, especially when blindness began early {cite:p}`Gougoux2004`. Brain imaging shows why: regions that would have processed vision are recruited for hearing, an example of the [neuroplasticity](the-brain.ipynb) that the next chapter describes {cite:p}`Voss2016`.
- **The method** compares small groups of early-blind, late-blind, and sighted listeners on specific tasks, which is why the findings are task-specific rather than a general verdict on hearing.
- **The limits** are the interesting part. The same early-blind listeners are *worse* at judging the elevation of a sound source {cite:p}`Zwiers2001`, because the cues for up and down are subtle spectral ones that the auditory system normally calibrates against vision. Without sight to teach it, that part of spatial hearing never fully develops. Enhancement in one task and deficit in another is not "better hearing"; it is hearing that has been reorganised.

The popular claim mistakes skill for sensation. Blind musicians play well for the reason sighted ones do, which is thousands of hours of practice, sometimes with attention that vision would otherwise have claimed. What these studies show is subtler and more relevant to this chapter: the senses calibrate each other, and when one is missing the others are re-tuned rather than simply turned up.
