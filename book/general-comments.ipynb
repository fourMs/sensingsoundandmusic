This page collects practical tips and tricks for doing well in MUS2640 and in university more generally: how to write and reference clearly, how to study effectively, and how to prepare for exams.

---

## Academic writing

### Understand the assignment
- Read the prompt and identify the **task** (compare, explain, argue, reflect, etc.).
- Note **word limits**, **deadline**, **format** (essay, short answers, portfolio), and which **citation style** to use. If anything is unclear, ask on the course forum or in class.

### Plan before you draft
- **Outline** your main line of argument in a few bullet points.
- Distinguish **description** (what something is) from **analysis** (how and why it matters in context).
- For each main point, decide what **evidence** you will use: reading, listening examples, or data from the course.

### Structure
- **Introduction:** context, the question you answer, and a **thesis** (your main claim) in one or two clear sentences.
- **Body:** one main idea per paragraph, with a **topic sentence**, evidence, and a short **link** to the next idea.
- **Conclusion:** synthesize (do not just repeat); you can state limitations or what would still need to be studied.

### Voice and style
- **Third person** is often best for analysis (“The results suggest…”, “This argument …”). **First person** is fine for methods, reflection, or when the assignment asks for your experience.
- Prefer **active** voice where it keeps the sentence clear.
- Be **consistent**: do not switch voice without a reason.

### Academic integrity
- Cite every idea that is not common knowledge and every **quotation**, **figure**, or **paraphrase** from a source.
- Keep your own notes separate from copied text so you never paste uncredited material by mistake.
- If your institution offers a **similarity/plagiarism** tool, use it as a *check*, not a substitute for careful referencing.

### Revision
- Read once for **structure** (does the argument flow?), once for **claims and evidence** (is each claim supported?), and once for **language** and **references**.

### Formatting basics (including when you write in Markdown)
- Separate paragraphs with a **blank line** between them.
- Leave a blank line before and after lists, headings, and block quotes so they render correctly.
- Avoid random single line breaks inside a paragraph; they break reading flow in exported PDFs and web pages.

## Referencing and source use

### In-text citations
- Every time you use someone else’s words, ideas, data, or a distinctive formulation, add an **in-text citation** in the style your instructor requires (author–year, numbered, footnotes, etc.).
- **Paraphrase** in your own words and still cite the source. **Quote** sparingly; when you quote, use quotation marks and a page number if the style asks for it.

### Reference list
- List **only** sources you actually cited in the text.
- Follow **one** style manual consistently (APA, Chicago, IEEE, etc.—whatever the course specifies).
- Alphabetize by first author’s surname (unless your style uses order of appearance).
- Check **spelling of names**, **years**, **titles**, **volume/issue/pages**, and **DOIs or URLs**. Reference managers (e.g. **Zotero**, **Mendeley**) help, but always **proofread** the exported list.

Here are APA-shaped patterns (adapt punctuation to your required style):

| Type | Pattern (illustrative) |
|------|------------------------|
| Book | Author, A. A. (Year). *Title of book*. Publisher. |
| Journal article | Author, A. A. (Year). Title of article. *Journal Title*, *volume*(issue), pages. https://doi.org/… |
| Chapter in edited book | Author, A. A. (Year). Title of chapter. In E. Editor (Ed.), *Title of book* (pp. xx–xx). Publisher. |
| Web page | Author or Organization. (Year, Month Day). Title of page. *Site name*. URL |

If no style is specified for a small task, ask—guessing is how inconsistent bibliographies happen.

## Study technique

### Time and attention
- **Short focused blocks** (e.g. 25–45 minutes) with breaks beat endless passive reading.
- **Reduce distractions** when you need deep work: silence notifications, or use a separate browser profile for study.

### Active learning
- After a lecture or chapter, **close the notes** and write a **short summary** in your own words.
- **Explain** a concept aloud or to a peer as if they were new to the topic (the Feynman-style check).
- Turn headings into **questions** and try to answer them before rereading.

### Notes and the course material
- Link new ideas to **examples** from sound, music, or demos in the book—concrete examples anchor abstract terms (masking, meter, timbre, etc.).
- If you use digital notes, **tag** topics (e.g. `pitch`, `rhythm`, `ethics`) so you can review before the exam.

### Spacing and variety
- **Revisit** material over several days instead of only cramming the night before—spacing improves retention.
- **Mix** topics in a study session (e.g. psychoacoustics + one musicology theme) once you know the basics; it builds flexible recall.

### Balance
- Steady sleep, food, and movement support memory more than marathon all-nighters.

## Exam preparation

### Know what you are preparing for
- Clarify **format**: multiple choice, short answers, essay, listening tasks, or a combination.
- Check whether **aids** are allowed (notes, dictionary, headphones) and **time** per section.

### Plan backward from the date
- List **topics** from the syllabus or learning outcomes.
- Rate each topic **green / yellow / red** for confidence and spend more time on yellow and red.

### Practice under constraints
- Answer **past papers** or invent questions from section headings with a **timer**.
- For essay-style exams, practice **outlines** (thesis + three supporting points) in 10–15 minutes.

### The last days
- Prefer **review and sleep** over trying to read everything once more from scratch.
- Prepare **logistics**: device charged, required software tested, room and time zone for digital exams.

### During the exam
- **Skim** the whole paper first; allocate time per section.
- For essays: **plan** briefly, then write; leave a few minutes to **check** names, concepts, and instructions (“compare” vs. “define”).
- If you draw a blank, **start** with what you know and circle back; partial credit usually beats an empty answer.
