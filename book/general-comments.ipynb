## Writing Academically

Writing an academic essay is a key skill for university students. Here are some essential guidelines to help you succeed:

### 1. Understand the Assignment
- **Read the prompt carefully.** Identify the main question or topic.
- **Clarify requirements.** Note word count, formatting, and referencing style.

### 2. Plan and Research
- **Brainstorm ideas.** Outline your main arguments.
- **Conduct research.** Use credible academic sources (books, journal articles, reputable websites).
- **Take notes.** Organize information by theme or argument.

### 3. Structure Your Essay
A typical academic essay includes:
- **Introduction:** Present the topic, provide background, and state your thesis (main argument).
- **Body Paragraphs:** Each paragraph should focus on one main idea, supported by evidence and analysis.
- **Conclusion:** Summarize your arguments and restate the thesis in light of the evidence.

### 4. Write Clearly and Objectively
- **Use formal language.** Avoid slang and contractions.
- **Be concise and precise.** Express ideas clearly and avoid unnecessary words.
- **Support claims with evidence.** Reference all sources appropriately.

### 5. Reference Properly
- **Cite all sources.** Use the required citation style (APA, MLA, Chicago, etc.).
- **Include a bibliography or reference list.**

### 6. Revise and Proofread
- **Check structure and flow.** Ensure each paragraph connects logically.
- **Edit for grammar and spelling.**
- **Verify citations and formatting.**

### 7. Academic Integrity
- **Avoid plagiarism.** Always acknowledge the work of others.
- **Use plagiarism checkers** if available.

---
### First or third person

- Check the assignment rubric and ask your instructor if unclear — their preference overrides general rules.
- Use third person (formal, impersonal voice) for most academic essays and analytic discussion: it creates objectivity and sounds more scholarly.
    - Example: "This essay argues that X influences Y."
- Use first person when describing your own actions, reflections, or decisions (methods, personal response, reflective pieces).
    - Example (methods): "I conducted three interviews..." or use "we" if a group or to avoid a lone-I.
- Prefer active voice for clarity, whether first or third person. Active + third person is usually best for analysis.
- Be consistent: choose one voice for the main body and only switch when genre or section requires it (e.g., methods or reflection).
- If unsure, default to third person or neutral constructions (e.g., "The results indicate...") — these are safe for most bachelor-level essays.

### How to add references

### How to make an academic reference list

- Follow the required citation style (APA, MLA, Chicago, IEEE, etc.). Your instructor or publisher’s guidelines override generic rules.
- Include only sources cited in the text. Do not list sources you consulted but did not cite.
- Order entries alphabetically by the first author’s last name (or by title for works with no author) unless the style specifies a different order (e.g., numbered citation styles).
- Use a hanging indent for each entry (first line flush left, subsequent lines indented).
- Be consistent with punctuation, capitalization, and italicization according to the chosen style.
- Provide complete and accurate bibliographic details: author(s), year, title, source (journal/book/publisher), volume/issue/pages for articles, DOI or stable URL for online resources.
- Prefer DOIs for journal articles; if none, include a stable URL and an access date only if the style requires it or the content is likely to change.
- Abbreviate multiple authors per the style’s rules (e.g., list all authors up to a limit, then use “et al.” if allowed).
- For chapters in edited volumes, include chapter author, year, chapter title, “In” editor(s), book title, page range, and publisher.
- Double-check: spelling of author names, year, title, page numbers, and DOI/URL. Small errors make sources hard to find.
- Use reference-management software (Zotero, Mendeley, EndNote, Paperpile) to store sources and format the list reliably; export to the required style and verify formatting manually.

Common examples (APA-style formatting guidance):
- Book: Author, A. A. (Year). Title of book (Edition if any). Publisher.
- Journal article: Author, A. A., & Author, B. B. (Year). Title of article. Title of Journal, volume(issue), pages. `https://doi.org/xx.xxx/yyyy`
- Chapter in edited book: Author, A. A. (Year). Title of chapter. In E. Editor & F. Editor (Eds.), Title of book (pp. xx–xx). Publisher.
- Webpage: Author, A. A. (Year, Month Day). Title of page. Site Name. URL

Final checks:
- Ensure the in-text citations match the reference list entries exactly.
- Run a style-specific validator or use your reference manager to catch missing fields and formatting issues.

### Line breaks and paragraph spacing

- Use a blank line (one empty line) to separate paragraphs. This is the standard and produces clear, consistent paragraph breaks in Markdown and most editors.
- Avoid inserting single hard line breaks within a paragraph to force visual spacing — they break flow and can cause inconsistent rendering across formats.
- If you prefer indentation for paragraphs, pick that convention and apply it consistently; do not mix blank-line separation and in-line indentation.
- For intentional line breaks within a paragraph (poetry, addresses, or controlled line breaks), use two spaces at the end of the line or an explicit HTML `<br>`; otherwise avoid.
- Leave a blank line before and after block elements (lists, code blocks, blockquotes, headings) to ensure correct formatting.

### Window types

When computing a spectrogram, the audio signal is divided into short overlapping segments, and each segment is multiplied by a mathematical function called a **window** before applying the Fourier transform. The window reduces edge effects and spectral leakage, improving the accuracy of the frequency analysis.

Common window types include:

- **Hanning (Hann) window:** Smoothly tapers the segment to zero at the edges, reducing spectral leakage. Widely used for audio analysis.
- **Hamming window:** Similar to Hann but with slightly different tapering, offering a good balance between main lobe width and side lobe suppression.
- **Blackman window:** Provides even better side lobe suppression, useful when minimizing leakage is critical.
- **Rectangular window:** No tapering (all ones); can cause significant spectral leakage and is rarely used in practice.
- **Terhardt window:** Designed by Ernst Terhardt, this window function is tailored for psychoacoustic analysis and auditory modeling. It emphasizes perceptual aspects of spectral analysis, aiming to better represent how the human ear perceives sound, though it is less commonly used in general audio processing.

The choice of window affects the trade-off between frequency resolution and leakage. In music and audio analysis, the Hann window is most commonly used due to its good overall performance.

For more on window functions, see [Wikipedia: Window function](https://en.wikipedia.org/wiki/Window_function).

```python
# Define Terhardt window function
def terhardt_window(N):
    """
    Create a Terhardt window for psychoacoustic analysis.
    This is a simplified version - adjust based on your needs.
    """
    n = np.arange(N)
    # Simple approximation of Terhardt window
    w = np.exp(-0.5 * ((n - N/2) / (N/8))**2)
    return w / np.max(w)


win_length = 1024

plt.figure(figsize=(16, 10))
for i, win in enumerate(window_types):
    # Get window array
    if win == 'terhardt':
        w = terhardt_window(win_length)
    else:
        w = np.hanning(win_length) if win == 'hann' else \
            np.hamming(win_length) if win == 'hamming' else \
            np.blackman(win_length) if win == 'blackman' else \
            np.ones(win_length)  # boxcar

    # Plot window
    plt.subplot(2, len(window_types), i + 1)
    plt.plot(w)
    plt.title(f"{win.capitalize()} window")
    plt.ylim(-0.1, 1.1)
    plt.xticks([])
    plt.yticks([])

    # Compute STFT with the given window
    D = librosa.stft(y, window=w, n_fft=win_length, hop_length=256)
    S_db = librosa.amplitude_to_db(np.abs(D), ref=np.max)

    # Plot spectrogram
    plt.subplot(2, len(window_types), i + 1 + len(window_types))
    librosa.display.specshow(S_db, sr=sr, x_axis='time', y_axis='hz', cmap='magma')
    plt.title(f"Spectrogram ({win.capitalize()})")
    plt.colorbar(format='%+2.0f dB')
    plt.tight_layout()

plt.show()
```
